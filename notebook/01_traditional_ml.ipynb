```python
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
import os

print(os.listdir("datasets"))
```

```python
with open("datasets/tamil_offensive_full_train.csv", "r", encoding="utf-8") as f:
    for _ in range(10):
        print(repr(f.readline()))
```

```python
train_df = pd.read_csv(
    "datasets/tamil_offensive_full_train.csv",
    sep="\t",
    header=None,
    names=["text", "label", "extra"]
)

dev_df = pd.read_csv(
    "datasets/tamil_offensive_full_dev.csv",
    sep="\t",
    header=None,
    names=["text", "label", "extra"]
)

test_df = pd.read_csv(
    "datasets/tamil_offensive_full_test.csv",
    sep="\t",
    header=None,
    names=["text", "label", "extra"]
)
```

```python
train_df.head()
```

```python
train_df.shape
```

```python
train_df["label"].value_counts()
```

```python
train_df = train_df.drop(columns=["extra"])
dev_df = dev_df.drop(columns=["extra"])
test_df = test_df.drop(columns=["extra"])
```

```python
train_df.head()
```

```python
X_train = train_df["text"]
y_train = train_df["label"]

X_dev = dev_df["text"]
y_dev = dev_df["label"]

X_test = test_df["text"]
y_test = test_df["label"]
```

```python
print(X_train.head())
```

```python
print(y_train.head())
```

```python
vectorizer = TfidfVectorizer()
```

```python
X_train_tfidf = vectorizer.fit_transform(X_train)

X_dev_tfidf = vectorizer.transform(X_dev)

X_test_tfidf = vectorizer.transform(X_test)
```

```python
print(X_train_tfidf.shape)
```

```python
print(X_train_tfidf.shape)
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train_tfidf, y_train)
```

```python
predictions = model.predict(X_dev_tfidf)
```

```python
print("Accuracy:", accuracy_score(y_dev, predictions))
print()
print(classification_report(y_dev, predictions))
```

```python
model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)
model.fit(X_train_tfidf, y_train)

predictions = model.predict(X_dev_tfidf)

print(classification_report(y_dev, predictions))
```

```python
# ==========================================
# Experiment 3 - Linear SVM
# ==========================================

from sklearn.svm import LinearSVC

svm_model = LinearSVC(random_state=42)

svm_model.fit(X_train_tfidf, y_train)

svm_predictions = svm_model.predict(X_dev_tfidf)

print("Accuracy:", accuracy_score(y_dev, svm_predictions))
print()
print(classification_report(y_dev, svm_predictions))
```

```python
vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)

X_dev_tfidf = vectorizer.transform(X_dev)

X_test_tfidf = vectorizer.transform(X_test)
```

```python
# Experiment 4 - Character TF-IDF + Linear SVM
```

```python
char_vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(3,5)
)
```

```python
X_train_char = char_vectorizer.fit_transform(X_train)

X_dev_char = char_vectorizer.transform(X_dev)

X_test_char = char_vectorizer.transform(X_test)
```

```python
print(X_train_char.shape)
```

```python
char_svm = LinearSVC(random_state=42)

char_svm.fit(X_train_char, y_train)
```

```python
char_predictions = char_svm.predict(X_dev_char)
```

```python
print(classification_report(y_dev, char_predictions))
```

```python
# Experiment 5 - Word + Character TF-IDF + Linear SVM
```

```python
from scipy.sparse import hstack
```

```python
X_train_combined = hstack([X_train_tfidf, X_train_char])

X_dev_combined = hstack([X_dev_tfidf, X_dev_char])

X_test_combined = hstack([X_test_tfidf, X_test_char])
```

```python
print(X_train_combined.shape)
```

```python
combined_svm = LinearSVC(random_state=42)

combined_svm.fit(X_train_combined, y_train)
```

```python
combined_predictions = combined_svm.predict(X_dev_combined)
```

```python
print(classification_report(y_dev, combined_predictions))
```

```python
# Experiment 6 - Hyperparameter Tuning with GridSearchCV
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "C": [0.1, 0.5, 1, 2, 5, 10]
}
```

```python
grid = GridSearchCV(
    LinearSVC(random_state=42),
    param_grid,
    cv=5,
    scoring="f1_weighted",
    n_jobs=-1
)
```

```python
grid.fit(X_train_char, y_train)
```

```python
print("Best Parameters:", grid.best_params_)
print("Best CV Score:", grid.best_score_)
```

```python
best_model = grid.best_estimator_

grid_predictions = best_model.predict(X_dev_char)

print(classification_report(y_dev, grid_predictions))
```

```python
import os
import joblib

os.makedirs("models", exist_ok=True)

joblib.dump(best_model, "models/char_svm.pkl")
joblib.dump(char_vectorizer, "models/char_vectorizer.pkl")

print("Model saved successfully!")
```
